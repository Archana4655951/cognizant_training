```python
import matplotlib.pyplot as plt

subjects = ["Math", "Science", "English", "Social", "Hindi"]
marks = [80, 75, 90, 85, 70]

plt.plot(subjects, marks, marker="o")
plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 200]

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jun", "Jul", "Aug", "Sep", "Oct"]
rainfall = [120, 180, 150, 100, 80]

plt.plot(months, rainfall, marker="o")
plt.title("Rainfall")
plt.xlabel("Months")
plt.ylabel("Rainfall (mm)")
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
temp = [30, 32, 31, 34, 33]

plt.plot(days, temp, marker="o")
plt.title("Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.show()
```
