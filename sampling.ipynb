```python
import random
clusters = {
    "Hyd" : [1,2,3,4,5],
    "Pune" :[6,7,8,9,10],
    "Goa" :[11,12,13,14,15]
}
select_clusters = ["Hyd","Goa"]
sample = []
for cluster in select_clusters:
    sample.extend(clusters[cluster])
    print("clusters sample:",sample)
```

```python
import random
clusters = ["class A","Class B","Class C","Class D"]
sample = random.choice(clusters)
print("Selected Clusters:",sample)
```

```python
employee = {
    "A" : 10,
    "B" : 20,
    "C" : 30,
    "D" : 40
}
sample =[name for name,emp in employee.items() if emp>=10]
print("Judgement Sample :",sample)
```

```python
print("Snow Ball sampling")
```

```python
networks = {
    "A" : ["B","C"],
    "B" : ["D"],
    "C" : ["E"]
}
sample = ["A"]
sample.extend(networks["A"])
sample.extend(networks["B"])
print("SnowBall sampling :",sample)
```
