```python
import matplotlib.pyplot as plt
plt.subplot(1,2,1)
plt.plot([1,2,3],[10,20,30])
plt.subplot(1,2,2)
plt.plot([1,2,3],[30,20,10])
plt.show()
```

```python
months = ["Jan","Feb","Mar"]
sales = [100,200,300]
profit = [20,40,60]
plt.plot(months,sales,label = "sales")
plt.plot(months,profit,label = "profit")
plt.legend()
plt.show()
```

```python
months = ["Jan","Feb","Mar"]
sales = [1000,2000,3000]
plt.plot(months,sales)
plt.show()
```

```python
months = ["Jan","Feb","Mar"]
sales = [13500,22360,39780]
plt.plot(months,sales,alpha = 0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1,9,3,4,1]
y1 = [2,3,9,5,6]
y2 = [1,3,2,7,1]

plt.subplot(1,2,1)
plt.plot(x,y1)
plt.title("plot 1")

plt.subplot(1,2,2)
plt.plot(x,y2)
plt.title("plot 2")

plt.show()

```
