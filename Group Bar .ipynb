```python
import matplotlib.pyplot as plt
import numpy as np
products =  ["Mobile","Laptop","Tv"]
sales_2024 = [2000,300,5000]
sales_2025 = [4000,2498,938]

x= np.arange(len(products))
width = 0.5
plt.bar(x-width/2,sales_2024,width,label='2024')
plt.bar(x+width/2,sales_2025,width,label='2025')
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales comparsion : 2024 vs 2025")
plt.xticks(x,products)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

students = ["A", "B", "C"]
math = [80, 75, 90]
science = [85, 70, 95]

x = np.arange(len(students))
width = 0.35

plt.bar(x-width/2, math, width, label="Math")
plt.bar(x+width/2, science, width, label="Science")

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Marks Comparison")
plt.xticks(x, students)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

classes = ["Class 1", "Class 2", "Class 3"]
boys = [20, 25, 18]
girls = [22, 24, 20]

x = np.arange(len(classes))
width = 0.4

plt.bar(x-width/2, boys, width, label="Boys")
plt.bar(x+width/2, girls, width, label="Girls")

plt.xlabel("Classes")
plt.ylabel("Students")
plt.title("Boys vs Girls")
plt.xticks(x, classes)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

products = ["Pen", "Book", "Bag"]
sales2024 = [100, 150, 120]
sales2025 = [120, 170, 140]

x = np.arange(len(products))
width = 0.35

plt.bar(x-width/2, sales2024, width, label="2024")
plt.bar(x+width/2, sales2025, width, label="2025")

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales Comparison")
plt.xticks(x, products)
plt.legend()
plt.show()
```
