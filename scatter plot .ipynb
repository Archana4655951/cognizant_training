```python
import matplotlib.pyplot as plt
one = [1,9,3,4,5]
two = [2,4,6,2,10]
plt.scatter(one,two)
plt.title("Scatter plot")
plt.xlabel("one")
plt.ylabel("tw0")
plt.show()
```

```python
import matplotlib.pyplot as plt

hours = [1,2,5,8,1]
marks = [37,23,39,12,11]

plt.scatter(hours,marks)
plt.title("Study hours vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.show()

```
