```python
import numpy as np
products=["bag","chair","bottle","fan","book"]
sales_2024=[50,100,80,100,150]
sales_2025=[70,140,100,105,200]
x=np.arange(len(products))
width=0.35
plt.bar(x-width/2,sales_2024,width)
plt.bar(x-width/2,sales_2025,width)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
products =["sarees","books","kettle"]
online  =[3000,2000,4000]
offline =[1500,1000,5000]
plt.bar(products,online)
plt.bar(products ,offline, bottom=online)
plt.show()
```

```python
import matplotlib.pyplot as plt
students=["sweety","archana","narukulla"]
sem1=[8.2,8.5,8.7]
sem2=[9.5,9.3,9.2]
plt.bar(students,sem1)
plt.bar(students,sem2,bottom=sem1)
plt.show()
```

```python
import matplotlib.pyplot as plt
Customers = list(range(0,100))
k =10
sample = Customers[::k]
print("Systematic_sample:",sample)
```

```python
import matplotlib.pyplot as plt
import random 
developer = list(range(1,50))
manager = list(range(1,10))
tester = list(range(81,100))

sample_dev = random.sample(developer,5)
sample_man = random.sample(manager,8)
sample_test = random.sample(tester,7)

final_sample = sample_dev + sample_man + sample_test
print(final_sample)
```

```python
import matplotlib.pyplot as plt
import numpy as np
subjects = ["Math","Science","English","Computer"]
marks = [85,90,78,95]

result = np.arange(len(subjects))
plt.bar(result,marks)

plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()

```
