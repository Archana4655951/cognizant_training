```python
import pandas as pd 
data={"Name":["sweety","kavita","archana"],
      "age":[19,30,20],
      "City":["knr","jgtl","hyd"],
     "grades":["A","B","C"]
}
details=pd.DataFrame(data)
print(details)
print(details.shape)
print(details.index)
print(details.head())
print(details.sample(2))
print(details.info())
print(details.describe())
print(details.iloc[0])
print(details.grades)
```

```python
import pandas as pd
dataJoin=pd.DataFrame({
    "ID":[4,5,6],
    "Name":["sweety","sravika","sathwka"]})
dataJoin2=pd.DataFrame({
    "ID":[1,2,3],
    "Name":["archana","radha","krishna"]})
print(dataJoin)
print(dataJoin2)
result=pd.merge(dataJoin,dataJoin2,on="ID")
print(result)
```

```python
duplicate = pd.DataFrame({
    "ID" :[1,2,3],
    "Name" :["sweety","archana","archana"],
    "Age" : [ 19,19,40]
})
print(duplicate.nunique)
print(duplicate)
print(duplicate.Name.duplicated())
print(duplicate.Age.duplicated())
print(duplicate.drop_duplicates())
```
