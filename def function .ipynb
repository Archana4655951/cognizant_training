```python
def square(n):
    return n*n 
result = square(6)
print(result)
```

```python
def cube(n):
    return n*n*n
result = cube(6)
print(result)
```

```python
def number(n):
    if n % 2 ==0:
        print("Even")
    else:
        print("Odd")
number(8)
number(7)
```

```python
def maximum(a,b):
    if a>b:
        print(a)
    else:
        print(b)

maximum(3,18)
maximum(17,7)
```

```python
def minimum(a,b):
    if a<b:
        print(a)
    else:
        print(b)
minimum(27,21)
minimum(37,12)
```
