```python
def fact(n):
    f = 1
    for i in range(1,n+1):
        f = f*i
    return f

number = 5
print(fact(number))
```

```python
def prime(n):
    count = 0

    for i in range(1,n+1):
        if n % i == 0:
            count = count +1
            
    if count == 2:
            return"Prime"
    else:
            return"Not Prime"

number = 7
print(prime(number))
```

```python
def count_vowels(s):
    count = 0

    for i in s:
        if i in "aeiou":
            count = count + 1
    return count
text = "python"
print(count_vowels(text))
```

```python
class Employee:
    def __init__(self,id,name,salary):
        self.id = id
        self.name = name
        self.salary = salary
    def display(self):
       print(self.id)
       print(self.name)
       print(self.salary)

e = Employee(77,"Sweety",40000)
e.display()
```
